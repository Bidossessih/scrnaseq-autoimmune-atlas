```python
import os
import numpy as np
import pandas as pd
import scanpy as sc
import anndata as an
import sys
sys.path.append('../')
from AIID_notebooks.inflapy.immunoFunc import *
```

```python
path = '/nfs/team205/bh14/Datasets/Remapped/raw_adata/'
```

```python
directory = ["Immune_compartment", "Immune_compartment_sorted_from_Tissue", "Tissue_or_Nonimmune"]
```

```python
files
```

```python
dirct = directory[0]
print("Processing: "+dirct)
files = os.listdir(path+"QC_Doublet/"+dirct)
files   
```

```python
for i in range(len(files)):
    print("... "+files[i])
        
    adata = sc.read_h5ad(path+"QC_Doublet/"+dirct+"/"+files[i])
        
    adata.obs["barcodes"] = adata.obs_names
        
    print(adata.shape)
        
    if 'adata_concat' in locals():
        adata_concat.obs_names_make_unique()
        adata_concat = adata_concat.concatenate(adata, join="inner", index_unique="_")
    else:
        adata_concat = adata
    print("Concatenated")    
    print(adata_concat.shape)
            
    adata_concat.write_h5ad(path+"QC_concatenated/"+dirct+"_Concatenated_dataset.h5ad")
    del adata
    
    gc.collect()
```
